```python
import json
def save_to_file(memory_list):
  with open('memories.json' ,'w') as f:
    json.dump(memory_list, f)
  print("Memories saved to file !")

def load_from_file():
  try:
    with open('memories.json' , 'r') as f :
      return json.load(f)
  except:
    return []
# Test it
memory = load_from_file()
memory.append("I am building a memory center")
save_to_file(memory)
```

```python
def add_memory(text):
  memories = load_from_file()
  memories.append(text)
  save_to_file(memories)
  print(f"Added: {text}")

def show_memories():
  memories = load_from_file()
  if not memories :
   print("No memories yet!")
  else:
    print("--- YOUR MEMORIES ---")
    for i, m in enumerate(memories):
      print(f"{i+1}. {m}")

#Test
add_memory("I am learning Python")
add_memory("I am applying to Ashoka University")
add_memory("I love South Indian Breakfast")
show_memories()
```

```python
def delete_memory(number) :
  memories = load_from_file()
  if 0 < number < len(memories):
    removed = memories.pop(number - 1)
    save_to_file(memories)
    print(f"Deleted: {removed} ")
  else:
    print("Invalid number")

# Test - show memories first then delete one
show_memories()
delete_memory(1)
show_memories()
```
